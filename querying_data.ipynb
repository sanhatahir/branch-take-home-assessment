## Notebook to ease querying the data

Simply put in your query in the 'select' text and run away! 

```python
import duckdb

with duckdb.connect("data/warehouse/branch.duckdb", read_only=True) as con:
    result = con.sql("""
        SELECT 
        
            rate_date, 
            source_currency, 
            destination_currency, 
            exchange_rate
            
        FROM mart.country_exchange_rates
        WHERE source_country_code = 'USA'
          AND destination_country_code = 'PAK'
          AND source_currency = 'USD'
          AND destination_currency = 'PKR'
        ORDER BY rate_date DESC
        LIMIT 3
    """).fetchall()

result
```
